```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
import numpy as np
import vedo
import trimesh
from trimesh.exchange import export
import open3d as o3d

# Define 3D coordinates
points = np.array([
    [0, 0, 0],
    [1, 2, 0],
    [2, 0, 1],
    [3, 1, 2],
    [4, 0, 0],
    [5, 1, 1]
])

# Initialize plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Function to plot a cube at a given position
def draw_cube(center, size=0.3, color='cyan'):
    r = size / 2
    x, y, z = center
    vertices = np.array([
        [x - r, y - r, z - r],
        [x + r, y - r, z - r],
        [x + r, y + r, z - r],
        [x - r, y + r, z - r],
        [x - r, y - r, z + r],
        [x + r, y - r, z + r],
        [x + r, y + r, z + r],
        [x - r, y + r, z + r]
    ])
    faces = [
        [vertices[i] for i in [0,1,2,3]],
        [vertices[i] for i in [4,5,6,7]],
        [vertices[i] for i in [0,1,5,4]],
        [vertices[i] for i in [2,3,7,6]],
        [vertices[i] for i in [1,2,6,5]],
        [vertices[i] for i in [4,7,3,0]]
    ]
    ax.add_collection3d(Poly3DCollection(faces, facecolors=color, linewidths=1, edgecolors='k', alpha=0.6))

# Loop through and draw
for i, pt in enumerate(points):
    size = 0.2 + 0.05 * i
    color = ['red', 'green', 'blue', 'orange', 'purple', 'yellow'][i % 6]
    draw_cube(pt, size=size, color=color)

# Set limits and show
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 3)
ax.set_zlim(-1, 3)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.tight_layout()
plt.show()

# Create a single 3D object — for example, a Sphere
sphere = vedo.Sphere(pos=[0, 0, 0], r=1.0, c='skyblue')

# Display the object (optional)
vedo.show(sphere, axes=1, title="Single Sphere")

# Export only this sphere to OBJ
vedo.write(sphere, "single_sphere.obj")
print("Exported single object to 'single_sphere.obj'")

# Create sphere data
phi, theta = np.mgrid[0.0:np.pi:100j, 0.0:2.0*np.pi:100j]
x = np.sin(phi) * np.cos(theta)
y = np.sin(phi) * np.sin(theta)
z = np.cos(phi)

# Plot
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='skyblue', alpha=0.8, edgecolor='k')

# Set aspect ratio and axes
ax.set_box_aspect([1, 1, 1])
ax.set_title("Sphere Visualization with Matplotlib")
plt.axis('off')
plt.show()

# Create a unit cube centered at the origin
cube = trimesh.creation.box(extents=(1, 1, 1))

# Export cube as STL binary format
stl_data = export.export_mesh(cube, 'cube_exported.stl', file_type='stl')

# Save to file
with open("cube_exported.stl", "wb") as f:
    f.write(stl_data)

cube = trimesh.creation.box(extents=(1, 1, 1))

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Extract the vertices and faces
verts = cube.vertices
faces = cube.faces

# Create the 3D polygon collection
mesh = Poly3DCollection(verts[faces], alpha=0.2)
mesh.set_facecolor('red')
ax.add_collection3d(mesh)

# Auto-scale to the mesh size
scale = verts.flatten()
ax.auto_scale_xyz(scale, scale, scale)

plt.show()

# Create a Cube using Open3D
cube = o3d.geometry.TriangleMesh.create_box(width=1.0, height=1.0, depth=1.0)

# Export Cube to OBJ file
o3d.io.write_triangle_mesh("cube.obj", cube)

# Load the Cube back using Open3D
cube = o3d.io.read_triangle_mesh("cube.obj")

# Get the vertices and faces of the cube
vertices = np.asarray(cube.vertices)
faces = np.asarray(cube.triangles)

# Visualize the Cube using Matplotlib
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Create the 3D polygon collection
mesh = Poly3DCollection(vertices[faces], alpha=0.5, facecolors='green')
ax.add_collection3d(mesh)

# Auto-scale to fit the mesh
scale = vertices.flatten()
ax.auto_scale_xyz(scale, scale, scale)

# Show the plot
plt.show()
```
