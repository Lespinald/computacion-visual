```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

uploaded = cv2.imread('/content/meowver1.png') 
image_rgb = cv2.cvtColor(uploaded, cv2.COLOR_BGR2RGB)

# RGB 
r, g, b = cv2.split(image_rgb)

# HSV 
image_hsv = cv2.cvtColor(uploaded, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(image_hsv)

# show RGB & HSV
fig, axs = plt.subplots(2, 3, figsize=(12, 6))
axs[0, 0].imshow(r, cmap='Reds'); axs[0, 0].set_title("Canal R")
axs[0, 1].imshow(g, cmap='Greens'); axs[0, 1].set_title("Canal G")
axs[0, 2].imshow(b, cmap='Blues'); axs[0, 2].set_title("Canal B")
axs[1, 0].imshow(h, cmap='hsv'); axs[1, 0].set_title("Canal H")
axs[1, 1].imshow(s, cmap='gray'); axs[1, 1].set_title("Canal S")
axs[1, 2].imshow(v, cmap='gray'); axs[1, 2].set_title("Canal V")
for ax in axs.flat:
    ax.axis('off')
plt.tight_layout()
plt.show()

# Slicing: 
image_mod = image_rgb.copy()
image_mod[50:350, 50:350] = [255, 0, 0]

# Swap
src_region = image_rgb[200:400, 200:400].copy()
image_mod[700:900, 700:900] = src_region

# Show
plt.imshow(image_mod)
plt.title("Swap n color")
plt.axis('off')
plt.show()

# Show histogram of intendity (RGB)
colors = ('r', 'g', 'b')
plt.figure(figsize=(8, 4))
for i, color in enumerate(colors):
    hist = cv2.calcHist([image_rgb], [i], None, [256], [0, 256])
    plt.plot(hist, color=color)
    plt.xlim([0, 256])
plt.title("Histograma de Intensidad - RGB")
plt.xlabel("Intensidad")
plt.ylabel("Frecuencia")
plt.show()

alpha = 1.2  # Contrast
beta = 50    # Brightness
image_brightness_contrast = cv2.convertScaleAbs(image_rgb, alpha=alpha, beta=beta)

plt.imshow(image_brightness_contrast)
plt.title(f"Contrast={alpha}   Brightness={beta}")
plt.axis('off')
plt.show()
```
