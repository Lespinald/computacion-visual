```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

uploaded = cv2.imread('/content/slynerd.jpg')
gray = cv2.cvtColor(uploaded, cv2.COLOR_BGR2GRAY)

# Binarize with fixed threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Find contours
contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw on copy
output = uploaded.copy()

# Loop through each contour
for i, cnt in enumerate(contours):

    cv2.drawContours(output, [cnt], -1, (0, 255, 0), 2)
    area = cv2.contourArea(cnt)
    perimeter = cv2.arcLength(cnt, True)

    # Moments and centroid
    M = cv2.moments(cnt)
    if M["m00"] != 0:
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
    else:
        cx, cy = 0, 0

    text = f"A:{int(area)} P:{int(perimeter)} C:({cx},{cy})"
    cv2.putText(output, text, (cx + 5, cy - 10), cv2.FONT_HERSHEY_SIMPLEX, 
                0.6, (255, 0, 0), 1, cv2.LINE_AA)

    # Mark centroid
    cv2.circle(output, (cx, cy), 3, (0, 0, 255), -1)

# Convert BGR to RGB
output_rgb = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)

# Show
plt.figure(figsize=(10, 6))
plt.imshow(output_rgb)
plt.title("Contours and Metrics")
plt.axis('off')
plt.tight_layout()
plt.show()
```
