```python
import trimesh
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

def load_model(path):
    mesh = trimesh.load_mesh(path)
    vertices = mesh.vertices
    faces = mesh.faces
    # Delete double edges
    unique_edges = set(frozenset(edge) for edge in mesh.edges)
    edges = np.array([list(e) for e in unique_edges])
    return vertices, edges, faces

def plot_model(vertices, edges, faces):
    fig = plt.figure()
    fig.set_size_inches(10, 10)
    ax = fig.add_subplot(projection='3d', box_aspect=(20, 60, 80))

    # Draw faces
    mesh = Poly3DCollection(vertices[faces], alpha=0.5)
    mesh.set_facecolor('#cccccc')
    mesh.set_edgecolor('red')
    ax.add_collection3d(mesh)

    # Draw vertices
    ax.scatter(vertices[:, 0], vertices[:, 1], vertices[:, 2], color='lime', s=5)


    ax.text2D(0.05, 0.95,
              f"# vértices: {vertices.shape[0]}\n# aristas: {edges.shape[0]}\n# caras: {faces.shape[0]}",
              transform=ax.transAxes,
              fontsize=12,
              verticalalignment='top',
              bbox=dict(facecolor='white', alpha=0.7))

    # Format
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")
    ax.view_init(elev=30, azim=45)
    ax.axis('off')

    plt.show()

vertices, edges, faces = load_model('cat.obj')
plot_model(vertices, edges, faces)
```
